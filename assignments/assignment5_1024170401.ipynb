## Q1: Write a program to create a NumPy 1D-array with 5 elements and perform basic operations like:

## Q2: Reverse the array arr = np.array([1, 2, 3, 6, 4, 5]). Find the most frequent value and their indices for:
- x = np.array([1,2,3,4,5,1,2,1,1,1])
- y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3])

````python
arr = np.array([1, 2, 3, 6, 4, 5])
print("Original Array:", arr)
print("Reversed Array:", arr[::-1])
````

**Answer:**
```
Original Array: [1 2 3 6 4 5]
Reversed Array: [5 4 6 3 2 1]
```

````python
x = np.array([1, 2, 3, 4, 5, 1, 2, 1, 1, 1])
unique_x, counts_x = np.unique(x, return_counts=True)
most_frequent_x = unique_x[np.argmax(counts_x)]
indices_x = np.where(x == most_frequent_x)[0]
print(f"Array x: {x}")
print(f"Most Frequent Element: {most_frequent_x}")
print(f"Indices: {indices_x}")
````

**Answer:**
```
Array x: [1 2 3 4 5 1 2 1 1 1]
Most Frequent Element: 1
Indices: [0 5 7 8 9]
```

````python
y = np.array([1, 1, 1, 2, 3, 4, 2, 4, 3, 3])
unique_y, counts_y = np.unique(y, return_counts=True)
most_frequent_y = unique_y[np.argmax(counts_y)]
indices_y = np.where(y == most_frequent_y)[0]
print(f"Array y: {y}")
print(f"Most Frequent Element: {most_frequent_y}")
print(f"Indices: {indices_y}")
````

**Answer:**
```
Array y: [1 1 1 2 3 4 2 4 3 3]
Most Frequent Element: 1
Indices: [0 1 2]
```

## Q5: Create array ucs420_AnhadKohli with values [10, 20, 30, 40, 50, 60, 70, 80, 90, 15, 20, 35]. Compute mean, median, max, min, unique elements. Reshape to 4x3 (reshaped_ucs420_AnhadKohli) and resize to 2x3 (resized_ucs420_AnhadKohli)

````python
ucs420_AnhadKohli = np.array([10, 20, 30, 40, 50, 60, 70, 80, 90, 15, 20, 35])
print(f"Mean: {np.mean(ucs420_AnhadKohli)}")
print(f"Median: {np.median(ucs420_AnhadKohli)}")
print(f"Max: {np.max(ucs420_AnhadKohli)}")
print(f"Min: {np.min(ucs420_AnhadKohli)}")
print(f"Unique Elements: {np.unique(ucs420_AnhadKohli)}")
````

**Answer:**
```
Mean: 43.333333333333336
Median: 37.5
Max: 90
Min: 10
Unique Elements: [10 15 20 30 35 40 50 60 70 80 90]
```

````python
reshaped_ucs420_AnhadKohli = ucs420_AnhadKohli.reshape(4, 3)
print("Reshaped (4x3):")
print(reshaped_ucs420_AnhadKohli)
````

**Answer:**
```
Reshaped (4x3):
```
